```python
# ponto de partida: o espectro de UMA imagem

import numpy as np
from PIL import Image


def crop_img(path):
    img = np.asarray(Image.open(path).convert("L"), dtype=np.float32) #abre a imagem e converte para a escala de cinza

    #recortar a imagem
    h, w = img.shape

    sobra_h = h - 250 #quanto sobra na vertical
    up = sobra_h // 2 #comeca na metade da sobra

    sobra_w = w - 250   #quanto sobra na horizontal
    side = sobra_w // 2 #comeca na metade da sobra

    crop = img[up : up + 250, side : side + 250]
    return crop

def spectrum_log(crop):
    F = np.fft.fftshift(np.fft.fft2(crop)) # frequência zero no centro
    return np.log1p(np.abs(F))        # escala log; mostre com plt.imshow


crop1 = crop_img("../data/test1.jpg")
crop2 = crop_img("../data/test2.png")

spec1 = spectrum_log(crop1)
spec2 = spectrum_log(crop2)
```

```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots(2, 2, figsize=(10, 10))   # 1 linha, 2 colunas de gráficos

ax[0, 0].imshow(crop1, cmap="gray");   ax[0, 0].set_title("Real: recorte")
ax[0, 1].imshow(spec1, cmap="inferno"); ax[0, 1].set_title("Real: espectro")
ax[1, 0].imshow(crop2, cmap="gray");   ax[1, 0].set_title("Gerada: recorte")
ax[1, 1].imshow(spec2, cmap="inferno"); ax[1, 1].set_title("Gerada: espectro")
plt.tight_layout()

plt.show()
```
